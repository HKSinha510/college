```python
import pandas as pd 
import matplotlib.pyplot as plt 
import numpy as np

df_student = pd.read_csv("dataset.csv", delimiter=',', header='infer')
```

```python
# study hour based on marks
# attendance based on marks


from scipy import stats

df_reg = df_student[['Marks', 'StudyHours']].copy()

m, b, r, p, se = stats.linregress(df_reg.Marks, df_reg.StudyHours)

df_reg['fx'] = (m * df_reg.Marks) + b
df_reg['error'] = df_reg.fx - df_reg.StudyHours


df_reg.plot.scatter(x='Marks', y='StudyHours')
plt.plot(df_reg.Marks, df_reg.fx)
plt.show()
print(df_reg)
```

```python
#import scipy

# study hour based on marks
# attendance based on marks

# Build model — Marks predicting StudyHours
m, b, r, p, se = stats.linregress(df_student['Marks'], df_student['StudyHours'])

print(f"Slope (m): {m:.4f}")
print(f"Intercept (b): {b:.4f}")
print(f"R² : {r**2:.4f}")
print(f"Equation: StudyHours = {m:.4f} * Marks + ({b:.4f})")
print(f"Standard Error: {se}")

# Making Predictions
test_marks = [24, 28, 50, 60, 80, 90]
for mark in test_marks:
    predicted = m * mark + b
    print(f"Marks = {mark} → Predicted StudyHours = {round(predicted, 2)}")
```

```python
#import scipy

# study hour based on marks
# attendance based on marks

# Build model — Marks predicting StudyHours
m, b, r, p, se = stats.linregress(df_student['Marks'], df_student['Attendance'])

print(f"Slope (m): {m:.4f}")
print(f"Intercept (b): {b:.4f}")
print(f"R² : {r**2:.4f}")
print(f"Equation: Attendance = {m:.4f} * Marks + ({b:.4f})")
print(f"Standard Error: {se}")

# Making Predictions
test_marks = [24, 28, 50, 60, 80, 90]
for mark in test_marks:
    predicted = m * mark + b
    print(f"Marks = {mark} → Predicted Attendance = {round(predicted, 2)}%")
```
