```python
import pandas as pd 
import matplotlib.pyplot as plt 
import numpy as np

df_student = pd.read_csv("dataset.csv", delimiter=',', header='infer')
```

```python
import seaborn as sns
sns.histplot(df_student['Marks'], kde=True, bins=6)
```

```python
sns.histplot(df_student['StudyHours'], kde=True, bins=6)
```

```python
sns.histplot(df_student['Attendance'], kde=True, bins=6)
```

```python
from scipy import stats
stats.probplot(df_student['Marks'], plot=plt)
```

```python
mean, std = df_student['Marks'].mean(), df_student['Marks'].std()
within_1sd = df_student[(df_student['Marks'] >= mean-std) & (df_student['Marks'] <= mean+std)].shape[0] / 30 * 100
within_2sd = df_student[(df_student['Marks'] >= mean - 2*std) & (df_student['Marks'] <= mean + 2*std)].shape[0] / 30 * 100  # ✓

print(within_1sd)
print(within_2sd)
```
