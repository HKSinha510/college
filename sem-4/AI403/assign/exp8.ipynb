```python
import pandas as pd 
import matplotlib.pyplot as plt 
import numpy as np

df_student = pd.read_csv("dataset.csv", delimiter=',', header='infer')
```

```python
from scipy.stats import pearsonr

pairs = [('Marks', 'StudyHours'), 
         ('Marks', 'Attendance'), 
         ('StudyHours', 'Attendance')]

for x, y in pairs:
    r, p = pearsonr(df_student[x], df_student[y])
    print(f"{x} vs {y}: r = {r:.4f}, p-value = {p:.4f}")
```

```python
import seaborn as sns
corr_matrix = df_student[['Marks', 'StudyHours', 'Attendance']].corr()
print(corr_matrix)
sns.heatmap(corr_matrix, annot=True, fmt='.4f')
```
